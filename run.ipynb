## Implement the Rag Chain

```python
class RagChain:
  def __init__(self, embedding_model, llm, vector_store, ans_prompt_templete, eval_prompt_templete):
    self.embedding_model = embedding_model
    self.llm = llm
    self.vector_store = vector_store
    self.ans_prompt_templete = ans_prompt_templete
    self.eval_prompt_templete = eval_prompt_templete

  def get_ans_prompt(self, context, query):
    return self.ans_prompt_templete.format(context=context, question=query)

  def get_eval_prompt(self, query, generated_response, ground_truth):
    return self.eval_prompt_templete.format(query=query, generated_response=generated_response, ground_truth=ground_truth)

  def get_contexts(self, query):
    contexts_with_score = self.vector_store.similarity_search_with_score(query, k=10)
    contexts = [doc.page_content for doc, score in contexts_with_score]
    return contexts

  def calculate_recall(self, context, retrieved_contexts):
    recall1 = context == retrieved_contexts[0]
    recall3 = context in retrieved_contexts[0:3]
    recall5 = context in retrieved_contexts[0:5]
    recall10 = context in retrieved_contexts[0:10]
    reciporcal_rank = 1 / (retrieved_contexts.index(context) + 1) if context in retrieved_contexts else 0
    return int(recall1), int(recall3), int(recall5), int(recall10), reciporcal_rank

  def calculate_bert_score(self, ground_truth, generated_answer, target_language):
    preds = [generated_answer]  # Tamil
    refs = [ground_truth]

    P, R, F1 = score(preds, refs, lang=target_language, model_type="xlm-roberta-base", verbose=True)
    return F1.mean()

  def calculate_llm_score(self, query, generated_answer, ground_truth, target_language):
    response = self.llm.invoke(self.get_eval_prompt(query, generated_answer, ground_truth))
    return response.replace(" ", "")

  def run(self, user_query, context, ground_truth, target_language):
    retrieved_contexts = self.get_contexts(user_query)
    recall1, recall3, recall5, recall10, reciporcal_rank = self.calculate_recall(context, retrieved_contexts)

    response = self.llm.invoke(self.get_ans_prompt(context="\n".join(retrieved_contexts[0:5]), query=user_query))
    bert_score = self.calculate_bert_score(ground_truth, response, target_language)
    llm_score = self.calculate_llm_score(user_query, response, ground_truth, target_language)

    return response, retrieved_contexts[0:5], recall1, recall3, recall5, recall10, reciporcal_rank, bert_score, llm_score
```

## Process data

```python
import time

def test(language, start_index, end_index):
  result = []
  for index, row in qa_data[start_index:end_index].iterrows():
    if index % 6 == 0 and index != 0:
      print("processing: ", index)
      time.sleep(90)
    try:
      response, contexts, recall1, recall3, recall5, recall10, reciporcal_rank, bert_score, llm_score = rag_chain.run(
        user_query=row[f"question_{language}"],
        context=row["context"],
        ground_truth=row[f"answer_{language}"],
        target_language=language
      )
      result.append({
          "context_id": row["id"],
          "context": row["context"],
          "question": row[f"question_{language}"],
          "ground_truth": row[f"answer_{language}"],
          "generated_response": response,
          "recall1": recall1,
          "recall3": recall3,
          "recall5": recall5,
          "recall10": recall10,
          "reciporcal_rank": reciporcal_rank,
          "bert_score": round(bert_score.item(), 2),
          "llm_score": float(llm_score)
      })
    except Exception as e:
      print(f"An error occurred: {e}")
      return result
  return result
```
